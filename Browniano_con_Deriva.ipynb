#Movimiento Browniano con Deriva
El movimiento browniano con deriva $X_t$ se define como:$$X_t = \mu t + \sigma B_t$$donde $B_t$ es un movimiento browniano estándar ($B_t \sim \mathcal{N}(0, t)$).
Por lo tanto, el valor del proceso en un tiempo $t$ sigue una distribución normal:$$X_t \sim \mathcal{N}(\mu t, \sigma^2 t)$$

Parámetros dados:

Deriva: $\mu = 0.6$Varianza instantánea: $\sigma^2 = 0.25 \implies \sigma = 0.5$Tiempo: $t = 4$Intervalo de interés: $[1, 3]$Calculamos la media y la varianza al tiempo $T = 4$:$$\mu_4 = \mu \cdot 4 = (0.6)(4) = 2.4$$$$\sigma^4_4 = \sigma^2 \cdot 4 = (0.25)(4) = 1.0 \implies \sigma_4 = 1.0$$

Buscamos la probabilidad:

$$
\begin{aligned}
P(1 < X_4 < 3) &= P\left(1 < 2.4 + \frac{1}{2} B_4 < 3\right) \\[6pt]
&= P\left(-1.4 < \frac{1}{2} B_4 < 0.6\right) \\[6pt]
&= P(-2.8 < B_4 < 1.2) \\[6pt]
&= \int_{-2.8}^{1.2} \frac{1}{\sqrt{2 * 4 * \pi}} e^{-\frac{x^2}{8}} \, dx \\[6pt]
&= 0.645
\end{aligned}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
import random as r
```

Movimiento Browniano:

```python
def browniano(T,L):
    lista = []
    w_0 = 0
    dt = T/L
    for _ in range(1,L):
        dw = r.gauss(0,1)
        w_1 = w_0 + np.sqrt(dt)*dw
        lista.append(w_1)

        w_0 = w_1
    lista.insert(0,0)
    return lista
```

```python
def deriva(mu, sigma, T, n):
  B = np.array(browniano(T, n))
  t = np.linspace(0, T, n)

  XT = mu * t + sigma * B

  return XT.tolist()
```

```python
def prob(mu, sigma, T, n, N):
  conteo = 0

  for _ in range(N):
    XT = deriva(mu, sigma, T, n)
    valor_f = XT[-1]

    if 1 <= valor_f <= 3:
      conteo += 1

  proba = conteo / N
  return print("La probabilidad de que el MB tome valores entre 1 y 3 al tiempo 4 es:",proba)
```
